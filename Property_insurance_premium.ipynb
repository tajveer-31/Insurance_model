```python
#Feature Preprocessing:
df['YEARBUILT'] = pd.to_datetime(df['YEARBUILT'], format='%Y')
date_formats = ['%d/%m/%Y', '%d-%m-%Y']
def convert_to_datetime(date_str):
    for fmt in date_formats:
        try:
            return pd.to_datetime(date_str, format=fmt)
        except ValueError:
            pass

df['P1_DOB'] = df['P1_DOB'].apply(convert_to_datetime)
df['COVER_START'] = df['COVER_START'].apply(convert_to_datetime)
df['QUOTE_DATE'] = df['QUOTE_DATE'].astype(str).str.strip().str.replace('/', '-')
df['QUOTE_DATE'] = pd.to_datetime(df['QUOTE_DATE'], format='%m-%d-%Y')
```

```python
#Model Training 
x= df.iloc[:,2:].drop('ANNUAL_PREMIUM', axis=1)
y= df['ANNUAL_PREMIUM']
```

```python
#Encoding the Categorical Variables(One-Hot Encoding
cols_to_encode = ['OCC_STATUS','P1_EMP_STATUS', 'P1_MAR_STATUS', 'P1_GENDER', 'PAYMENT_METHOD']

# Perform one-hot encoding for each column
for col in cols_to_encode:
    dummies = pd.get_dummies(df[col], prefix=col)
    x = x.drop(col, axis=1)
    x = pd.concat([x, dummies], axis=1)
```

```python
x.info()
```

```python
#Standardizing the Numerical features
from sklearn.preprocessing import StandardScaler
numerical_cols = x.select_dtypes(include = ['int64', 'float64']).columns
scaler = StandardScaler()
x[numerical_cols]= scaler.fit_transform(x[numerical_cols])
```

```python
x.head()
```

```python
#Split the Train and Test Data
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,train_size=0.8, random_state = 42)
```

Decision Tree

```python
from sklearn.tree import DecisionTreeRegressor
model = DecisionTreeRegressor(random_state=42)
model.fit(x_train, y_train)
y_test_pred = model.predict(x_test)

```

```python
# Evaluate the model
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, root_mean_squared_error
mae = mean_absolute_error(y_test, y_test_pred)
mse = mean_squared_error(y_test, y_test_pred)
rmse = root_mean_squared_error(y_test, y_test_pred)
r2 = r2_score(y_test, y_test_pred)

print("Decision Tree Metrics:")
print("Mean Absolute Error:", mae)
print("Mean Squared Error:", mse)
print("Root Mean Squared Error:", rmse)
print("R^2 Score:", r2)
```

```python
y_train_pred = model.predict(x_train)
print("Training R²:", r2_score(y_train, y_train_pred))

# Testing R-squared
print("Testing R²:", r2_score(y_test, y_test_pred))
```

Support Vector Regression


```python
from sklearn.svm import SVR
svr_model = SVR(kernel='rbf')
svr_model.fit(x_train, y_train)
y_pred_svr = svr_model.predict(x_test)
```

```python
mae_svr = mean_absolute_error(y_test, y_pred_svr)
mse_svr = mean_squared_error(y_test, y_pred_svr)
rmse_svr = root_mean_squared_error(y_test, y_pred_svr, squared=False)
r2_svr = r2_score(y_test, y_pred_svr)

print("Support Vector Regression (SVR) Metrics:")
print("Mean Absolute Error:", mae_svr)
print("Mean Squared Error:", mse_svr)
print("Root Mean Squared Error:", rmse_svr)
print("R^2 Score:", r2_svr)
```
